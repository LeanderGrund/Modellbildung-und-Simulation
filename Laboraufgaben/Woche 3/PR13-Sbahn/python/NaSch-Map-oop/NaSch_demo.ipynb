```python
import numpy as np
import matplotlib.pyplot as plt

import nasch
```

## Road class

```python
road = nasch.Road()

fig = plt.figure()
road.draw()

for k in range(2, 51):
    road.update()
    road.draw()

plt.close(fig)
```

## Map class (simple examples)

Node and road indices are 0-based.

```python
node1 = np.array([    0     ,   2   ,  3  ,   4   ,  5   ,  6  ,  7  ,  7      ])
node2 = np.array([    1     ,   3   ,  2  ,   6   ,  6   ,  7  ,  8  ,  9      ])
times = np.array([    2     ,   0   ,  0  ,   2   ,  2   ,  0  ,  1  ,  1      ])
nodex = np.array([ -1 , +1  ,  -1   , +1  ,  -1   , -1   , -.5 , +.5 , +1 , +1 ]) * 100
nodey = np.array([ +1 , +1  ,  -1   , -1  , -.5   , +.5  ,  0  ,  0  , -.5, +.5]) * 100
#                 \_________/ \___________/ \___________________________________/
#                   one-way      two-way                Split+Join
#                   2 nodes      2 nodes                6 nodes
#                   1 street     2 streets              5 streets

map = nasch.Map(node1=node1, node2=node2, nodex=nodex, nodey=nodey, times=times)

fig = plt.figure()
map.draw()

for k in range(2, 301):
    map.update()
    map.draw()

plt.close(fig)
```
