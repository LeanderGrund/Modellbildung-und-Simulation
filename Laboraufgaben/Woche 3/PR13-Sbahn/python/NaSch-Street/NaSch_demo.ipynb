Nagel Schreckenberg - periodic street

```python
import numpy as np
import matplotlib.pyplot as plt

from myInit import myInit
from myUpdate import myUpdate
from myDrawnow import myDrawnow
from TimeSpacePlot import TimeSpacePlot
```

 initialization

```python
u, p = myInit(plot=2)
```

time evolution

```python
t = 1
G = [u.copy()]
h = myDrawnow(u, p, t, None)

while t < p.tend:
    u = myUpdate(u, p)
    t = t + 1
    myDrawnow(u, p, t, h)
    G.append(u.copy())

# the rows of G are the street states at successive time steps
G = np.array(G)
plt.close(h.fig)  # the animation above is the output of this cell
```

sanity check

```python
s = np.count_nonzero(~np.isnan(G), axis=1)

if np.all(s == s[0]):
    print(f'Number of cars = {s[0]}')
else:
    raise RuntimeError('number of cars must not change')
```
